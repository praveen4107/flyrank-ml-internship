## 2. The question: decision, action, cost of a wrong call

Research question

Which content pages should be reviewed first for content refresh based on observable search and engagement signals?

Decision

The SEO or content team must decide which pages deserve review before others.

Unit of analysis

One content page (one row represents one page).

Output

A ranked refresh opportunity score with supporting reason codes.

Action

The highest-ranked pages will be reviewed for updating, expanding, improving CTR, or monitoring.

Cost of a wrong recommendation

High-potential pages may be ignored, resulting in lost organic traffic.

Editors may waste time refreshing pages that have little opportunity.

Low-quality prioritization reduces the effectiveness of limited review capacity.

Why ML helps

The problem is not to replace human judgment. ML helps combine multiple observable signals—such as impressions, CTR, position, freshness, and engagement—into a better ranking than a fixed rule alone.
