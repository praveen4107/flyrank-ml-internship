## 2. Target or proxy

The starter dataset provides trend_direction, which can be used as a proxy for identifying declining pages. I will define the target as:

Positive (1): trend_direction == "down"

Negative (0): all other pages

This is a proxy label for learning purposes and does not represent a guaranteed future decline.

```python
import pandas as pd
import os

# Clone the repository to access the data
!git clone https://github.com/praveen4107/flyrank-ml-internship.git

repo = "flyrank-ml-internship"
file_path = os.path.join(repo, "data", "raw", "content_refresh_anonymized.csv")

df = pd.read_csv(file_path)

df["target_declining"] = (df["trend_direction"] == "down").astype(int)

df[["content_id","client_id","trend_direction","target_declining"]].head()
```

An ML scoring approach can combine these observable signals into a more effective ranking while still supporting human review through transparent reason codes.
